```python
import os
import random
from PIL import Image
import matplotlib.pyplot as plt

# Function to extract and plot random images from all subdirectories
def display_random_images_from_all_dirs(base_dir, num_images=4):
    """
    Extract and display random images from each subdirectory of the given base directory.

    Parameters:
    - base_dir: The base directory containing subdirectories of images (e.g., Parasitized, Uninfected).
    - num_images: Number of random images to extract and display per subdirectory.
    """
    # List all subdirectories in the base directory
    subdirectories = [d for d in os.listdir(base_dir) if os.path.isdir(os.path.join(base_dir, d))]

    print(f"Extracting {num_images} random images from each subdirectory of: {base_dir}\n")

    for subdir in subdirectories:
        subdir_path = os.path.join(base_dir, subdir)
        images = os.listdir(subdir_path)
        images = [img for img in images if img.endswith(('.png', '.jpg', '.jpeg'))]  # Filter valid images

        print(f"Subdirectory: {subdir} (Total Images: {len(images)})\n")

        # Randomly select images
        selected_images = random.sample(images, min(num_images, len(images)))

        # Plot selected images
        fig, axes = plt.subplots(1, len(selected_images), figsize=(4 * len(selected_images), 4))
        fig.suptitle(f"Random Images from '{subdir}'", fontsize=12)

        for i, image_name in enumerate(selected_images):
            image_path = os.path.join(subdir_path, image_name)
            try:
                with Image.open(image_path) as img:
                    img.verify()  # Check if image is valid
                    img = Image.open(image_path)  # Re-open image for display
                    # Print image information
                    print(f"  Image {i+1}:")
                    print(f"    - Name: {image_name}")
                    print(f"    - Format: {img.format}")
                    print(f"    - Size: {img.size} (Width x Height)")
                    print(f"    - Mode: {img.mode}\n")
                    # Plot the image
                    axes[i].imshow(img)
                    axes[i].set_title(f"Image {i+1}")
                    axes[i].axis('off')
            except Exception as e:
                print(f"Error loading image {image_path}: {e}")
                continue

        plt.tight_layout()
        plt.show()

# Paths for training and test data
train_dir = "/content/cell_images/train"
test_dir = "/content/cell_images/test"

# Apply the function to both the train and test directories
print("Random Images from Training Directory")
display_random_images_from_all_dirs(train_dir, num_images=4)

print("\nRandom Images from Testing Directory")
display_random_images_from_all_dirs(test_dir, num_images=4)
```

**Observations:**

```python
# Define target image size
IMG_HEIGHT, IMG_WIDTH, CHANNELS = 64, 64, 3

# Initialize lists for images and labels
train_images = []
train_labels = []
test_images = []
test_labels = []
```

```python
#Define paths for the training and test data
train_dir = "/content/cell_images/train"
test_dir = "/content/cell_images/test"
# Image size for resizing
IMG_SIZE = 64

def preprocess_images(base_dir):
    images = []
    labels = []

    # Subdirectories: 'parasitized' and 'uninfected'
    categories = ['parasitized', 'uninfected']

    for category in categories:
        category_path = os.path.join(base_dir, category)
        label = 1 if category == 'parasitized' else 0

        for img_name in os.listdir(category_path):
            img_path = os.path.join(category_path, img_name)

            try:
                # Read the image using cv2
                img = cv2.imread(img_path)

                if img is not None:
                    # Resize the image to IMG_SIZE x IMG_SIZE
                    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))

                    # Append the image and label
                    images.append(img)
                    labels.append(label)
                else:
                    print(f"Skipping {img_name}: Unable to load.")
            except Exception as e:
                print(f"Error processing {img_name} in {category}: {e}")

    # Convert to NumPy arrays
    images = np.array(images, dtype=np.float32)

    print(f"Processed {len(images)} images from {base_dir}")
    return images, np.array(labels)

```

```python
# Preprocess training and test data
train_images, train_labels = preprocess_images(train_dir)
test_images, test_labels = preprocess_images(test_dir)
```

### <b>Converting RGB to HSV of Images using OpenCV

```python
# Function to convert all images in the dataset to HSV
def convert_to_hsv(images):
    hsv_images = []
    for img in images:
        # Convert each image to HSV
        hsv_image = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
        hsv_images.append(hsv_image)

    return np.array(hsv_images)
```

```python
def normalize_hsv_images(hsv_images, hue_range=360):
    """
    Normalize HSV images to the range [0, 1].

    Args:
    - hsv_images (numpy array): Array of HSV images with shape (batch_size, height, width, 3).
    - hue_range (int): Maximum range of the Hue channel (179 for OpenCV, 360 for other libraries).

    Returns:
    - hsv_normalized_images (numpy array): Array of normalized HSV images.
    """
    # Convert to float32 to ensure proper division
    hsv_images = hsv_images.astype('float32')

    # Normalize each channel
    hsv_images[:, :, :, 0] /= hue_range  # Normalize Hue
    hsv_images[:, :, :, 1] /= 255.0      # Normalize Saturation
    hsv_images[:, :, :, 2] /= 255.0      # Normalize Value

    return hsv_images

# Normalize train and test HSV images
train_images_hsv_normalized = normalize_hsv_images(train_images_hsv, hue_range=360)
test_images_hsv_normalized = normalize_hsv_images(test_images_hsv, hue_range=360)

# Verify the normalization
print("Train Images HSV - Min:", train_images_hsv_normalized.min(),
      "Max:", train_images_hsv_normalized.max(),
      "Data Type:", train_images_hsv_normalized.dtype)

print("Test Images HSV - Min:", test_images_hsv_normalized.min(),
      "Max:", test_images_hsv_normalized.max(),
      "Data Type:", test_images_hsv_normalized.dtype)
```

**Model Architecture Summary**

```python
def CNN_1(input_shape=(64, 64, 3), num_classes=2):
    """
    Create a CNN model for binary classification.

    Args:
    - input_shape (tuple): Shape of the input images, default is (64, 64, 3).
    - num_classes (int): Number of output classes, default is 2 (for binary classification).

    Returns:
    - model: Compiled Keras Sequential model.
    """
    model = Sequential()

    # Convolutional Block 1
    model.add(Conv2D(32, (2, 2), activation='relu', input_shape=input_shape))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 2
    model.add(Conv2D(32, (2, 2), activation='relu'))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 3
    model.add(Conv2D(32, (2, 2), activation='relu'))
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Flatten Layer
    model.add(Flatten())

    # Fully Connected Layer
    model.add(Dense(512, activation='relu'))
    model.add(Dropout(0.4))

    # Output Layer (Softmax for multi-class classification)
    model.add(Dense(num_classes, activation='softmax'))

    # Compile the model
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model
```

```python
import pandas as pd

def save_performance_summary(model_name, history, test_loss, test_accuracy, filename='cnn_performance.csv'):
  """
  Saves the training history and test performance of multiple Keras models to a CSV file
  and returns a Pandas DataFrame summarizing the results. Appends new rows for each model.

  Args:
    model_name (str): The name of the model (e.g., 'CNN_Model_1').
    history (keras.callbacks.History): The history object returned by model.fit().
    test_loss (float): The loss value on the test set.
    test_accuracy (float): The accuracy value on the test set.
    filename (str): The name of the CSV file to store the results (default: 'cnn_performance.csv').

  Returns:
    pandas.DataFrame: A DataFrame summarizing the models' performance.
  """

  # Create a dictionary to store the performance metrics
  performance = {
      'model_name': model_name,
      'train_loss': history.history['loss'][-1],
      'train_accuracy': history.history['accuracy'][-1],
      'val_loss': history.history['val_loss'][-1],
      'val_accuracy': history.history['val_accuracy'][-1],
      'test_loss': test_loss,
      'test_accuracy': test_accuracy
  }

  # Try to load existing data from the CSV file
  try:
    df = pd.read_csv(filename)
  except FileNotFoundError:
    df = pd.DataFrame(columns=performance.keys())  # Create an empty DataFrame if the file doesn't exist

  # Append the new performance data as a row
  df = pd.concat([df, pd.DataFrame([performance])], ignore_index=True)

  # Save the updated DataFrame back to the CSV file
  df.to_csv(filename, index=False)

  return df
```

```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix

def save_classification_report(model, model_name, test_images, test_labels_one_hot, filename='cnn_performance_summary.csv'):
  """
  Evaluates a CNN model on the test set, generates a classification report and confusion matrix,
  and saves the results to a Pandas DataFrame.

  Args:
    model: The trained Keras CNN model.
    model_name (str): The name of the model (e.g., 'CNN_Model_1').
    test_images: The NumPy array of test images.
    test_labels_one_hot: The NumPy array of one-hot encoded test labels.
    filename (str): The name of the CSV file to store the results (default: 'cnn_performance_summary.csv').

  Returns:
    pandas.DataFrame: A DataFrame summarizing the model's performance.
  """

  # Step 1 & 2: Generate predictions and true classes
  predictions = model.predict(test_images)
  predicted_classes = np.argmax(predictions, axis=1)
  true_classes = np.argmax(test_labels_one_hot, axis=1)

  # Step 3: Generate the classification report
  report = classification_report(true_classes, predicted_classes, target_names=["Uninfected", "Parasitized"], output_dict=True)

  # Step 4: Generate the confusion matrix
  conf_matrix = confusion_matrix(true_classes, predicted_classes)

  # Extract relevant metrics from the classification report
  performance = {
      'model_name': model_name,
      'accuracy': report['accuracy'],
      'uninfected_precision': report['Uninfected']['precision'],
      'uninfected_recall': report['Uninfected']['recall'],
      'uninfected_f1-score': report['Uninfected']['f1-score'],
      'parasitized_precision': report['Parasitized']['precision'],
      'parasitized_recall': report['Parasitized']['recall'],
      'parasitized_f1-score': report['Parasitized']['f1-score'],
      'true_negative': conf_matrix[0, 0],
      'false_positive': conf_matrix[0, 1],
      'false_negative': conf_matrix[1, 0],
      'true_positive': conf_matrix[1, 1]
  }

  # Try to load existing data from the CSV file
  try:
    df = pd.read_csv(filename)
  except FileNotFoundError:
    df = pd.DataFrame(columns=performance.keys())

  # Append the new performance data as a row
  df = pd.concat([df, pd.DataFrame([performance])], ignore_index=True)

  # Save the updated DataFrame back to the CSV file
  df.to_csv(filename, index=False)

  return df
```

```python
def CNN_Model_2_BN(input_shape=(64, 64, 3), num_classes=2):
    """
    Create a CNN model for binary classification with an additional convolutional layer
    and Batch Normalization.

    Args:
    - input_shape (tuple): Shape of the input images, default is (64, 64, 3).
    - num_classes (int): Number of output classes, default is 2 (for binary classification).

    Returns:
    - model: Compiled Keras Sequential model.
    """
    model = Sequential()

    # Convolutional Block 1
    model.add(Conv2D(32, (2, 2), activation='relu', input_shape=input_shape))
    model.add(BatchNormalization()) # Batch Normalization added here
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 2
    model.add(Conv2D(32, (2, 2), activation='relu'))
    model.add(BatchNormalization()) # Batch Normalization added here
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 3
    model.add(Conv2D(32, (2, 2), activation='relu'))
    model.add(BatchNormalization()) # Batch Normalization added here
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 4
    model.add(Conv2D(64, (2, 2), activation='relu'))
    model.add(BatchNormalization()) # Batch Normalization added here
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Flatten Layer
    model.add(Flatten())

    # Fully Connected Layer
    model.add(Dense(512, activation='relu'))
    model.add(BatchNormalization()) # Batch Normalization added here
    model.add(Dropout(0.4))

    # Output Layer
    model.add(Dense(num_classes, activation='softmax'))

    # Compile the model
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model
```

### **Model 3 with BatchNormalization and LeakyReLU as activation function**

```python
def cnn_Model_3_LRelu(input_shape=(64, 64, 3), num_classes=2):
    """
    Create a CNN model for binary classification using Batch Normalization
    and LeakyReLU activation.

    Args:
    - input_shape (tuple): Shape of the input images, default is (64, 64, 3).
    - num_classes (int): Number of output classes, default is 2 (for binary classification).

    Returns:
    - model: Compiled Keras Sequential model.
    """
    model = Sequential()

    # Convolutional Block 1
    model.add(Conv2D(32, (2, 2), input_shape=input_shape))
    model.add(BatchNormalization())
    model.add(LeakyReLU(alpha=0.1))  # LeakyReLU activation
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 2
    model.add(Conv2D(32, (2, 2)))
    model.add(BatchNormalization())
    model.add(LeakyReLU(alpha=0.1))  # LeakyReLU activation
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 3
    model.add(Conv2D(32, (2, 2)))
    model.add(BatchNormalization())
    model.add(LeakyReLU(alpha=0.1))  # LeakyReLU activation
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Convolutional Block 4
    model.add(Conv2D(64, (2, 2)))
    model.add(BatchNormalization())
    model.add(LeakyReLU(alpha=0.1))  # LeakyReLU activation
    model.add(MaxPooling2D(pool_size=(2, 2)))
    model.add(Dropout(0.2))

    # Flatten Layer
    model.add(Flatten())

    # Fully Connected Layer
    model.add(Dense(512))
    model.add(BatchNormalization())
    model.add(LeakyReLU(alpha=0.1))  # LeakyReLU activation
    model.add(Dropout(0.4))

    # Output Layer
    model.add(Dense(num_classes, activation='softmax'))

    # Compile the model
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model
```

```python
def CNN_augmented_model(input_shape=(64, 64, 3), num_classes=2):
    """
    Build a CNN model with LeakyReLU activation, BatchNormalization, Dropout, and Global Average Pooling.
    """
    model = Sequential()

    # Convolutional Block 1
    model.add(Conv2D(32, (3, 3), padding='same', kernel_initializer='he_normal', input_shape=input_shape))
    model.add(LeakyReLU(alpha=0.1))  # ReLU replaced with LeakyReLU
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.3))  # Dropout added to reduce overfitting

    # Convolutional Block 2
    model.add(Conv2D(64, (3, 3), padding='same', kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.4))  # Dropout added to reduce overfitting

    # Convolutional Block 3
    model.add(Conv2D(128, (3, 3), padding='same', kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.4))  # Dropout added to reduce overfitting

    # Convolutional Block 4 (optional deeper block)
    model.add(Conv2D(256, (3, 3), padding='same', kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.5))  # Dropout added to reduce overfitting

    # Global Average Pooling
    model.add(GlobalAveragePooling2D())  # Replaces Flatten to reduce overfitting

    # Fully Connected Layer
    model.add(Dense(256, kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))  # Regularization

    # Output Layer
    model.add(Dense(num_classes, activation='softmax'))  # Final output for classification

    # Compile the model
    model.compile(optimizer=Adam(learning_rate=0.001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model
```

```python
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D, BatchNormalization, LeakyReLU

def CNN_vgg16(input_shape=(64, 64, 3), num_classes=2):
    """
    Integrate VGG16 with custom CNN layers (fixed pooling issue).
    """
    # Load the VGG16 base model
    vgg_base = VGG16(weights='imagenet', include_top=False, input_shape=input_shape)

    # Freeze the VGG16 layers
    for layer in vgg_base.layers:
        layer.trainable = False

    # Custom layers
    model = Sequential()
    model.add(vgg_base)

    # Replace Flatten with Global Average Pooling
    model.add(GlobalAveragePooling2D())  # Avoid spatial dimension issues

    # Fully connected layers
    model.add(Dense(256, kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))

    model.add(Dense(128, kernel_initializer='he_normal'))
    model.add(LeakyReLU(alpha=0.1))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))

    # Output layer
    model.add(Dense(num_classes, activation='softmax'))

    # Compile the model
    model.compile(optimizer=Adam(learning_rate=0.001),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model

```

### **Building the Model**

```python
# Build the model
CNN_vgg16 = CNN_vgg16(input_shape=(64, 64, 3), num_classes=2)

# View the summary
CNN_vgg16.summary()
```

<b> using callbacks

```python
# Define callbacks
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
    tf.keras.callbacks.ModelCheckpoint(filepath='CNN_vgg16_best.keras', monitor='val_loss', save_best_only=True, verbose=1),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-6, verbose=1)
]
```

<b>Fit and Train the model

```python
# Train the model
history = CNN_vgg16.fit(
    train_generator,
    validation_data=val_generator,
    epochs=30,
    callbacks=callbacks,
    verbose=1
)
```
